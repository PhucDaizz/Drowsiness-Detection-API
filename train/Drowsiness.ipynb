```python
# Thêm vào cuối code trong Colab
from google.colab import files
import os

# Tìm file tflite
tflite_files = []
for root, dirs, filenames in os.walk('.'):
    for filename in filenames:
        if filename.endswith('.tflite'):
            tflite_files.append(os.path.join(root, filename))

# Download file tflite
if tflite_files:
    print(f"\nĐang download {tflite_files[0]}...")
    files.download(tflite_files[0])
else:
    print("Không tìm thấy file .tflite")

# Download labels.txt
if os.path.exists('labels.txt'):
    print("Đang download labels.txt...")
    files.download('labels.txt')
```
